# Question One

```python
text = input("Enter a string containing letters and numbers: ")

letters = []
numbers = []

for char in text:
    if char.isalpha():
        letters.append(char)
    elif char.isdigit():
        numbers.append(int(char))

word = "".join(letters)
total = sum(numbers)

print("Letters:", letters)
print("Numbers:", numbers)
print('Word:', word)
print("Sum:", total)
```

# Question Three

```python
price = float(input("Enter the total price: "))
member = input("Are you a member? (Yes/No): ").lower()

if price < 50:
    discount = 0
elif price <= 100:
    discount = 0.10
else:
    discount = 0.20

total = price - (price * discount)

if member == "yes":
    total = total - (total * 0.05)

print(f"Final Total: ${total:.2f}")
```

# Question Six

```python
def interleave(word1, word2):
    return "".join(a + b for a, b in zip(word1, word2))

word1 = input("Enter the first word: ")
word2 = input("Enter the second word: ")

if len(word1) != len(word2):
    print("Error: the words must be the same length.")
else:
    print(interleave(word1, word2))
```

# Question Seven

```python
def digit_sum(n):
    return sum(int(d) for d in str(n))

num = input("Enter a multi-digit number: ").strip()

if not num.isdigit() or len(num) < 2:
    print("Error: please enter a number with at least two digits.")
else:
    steps = []
    current = num
    while len(current) > 1:
        total = digit_sum(current)
        steps.append(f"{'+'.join(current)}={total}")
        current = str(total)

    print(f"Digital Root: {current} ({', '.join(steps)})")
```

# Question Eight

```python
def frequencies(numbers):
    counts = {}
    for n in numbers:
        counts[n] = counts.get(n, 0) + 1
    return list(counts.items())

nums = [1, 1, 2, 3, 3, 3, 4]
print(frequencies(nums))
```

# Question Nine

```python
n = int(input("Enter a number n: "))

if n < 1:
    print("Error: n must be at least 1.")
else:
    factorial = 1
    for i in range(1, n + 1):
        factorial *= i

    total = n * (n + 1) // 2

    print(f"Factorial is {factorial}, Sum is {total}.", end=" ")
    if factorial % total == 0:
        print(f"{factorial} is divisible by {total}.")
    else:
        print(f"{factorial} is not divisible by {total}.")
```

# Question Ten

```python
store = {'apple': 0.5, 'bread': 2.0, 'milk': 1.5}
cart = []

print("Store items:", ", ".join(f"{item} (${price})" for item, price in store.items()))
print("Type commands (e.g. Add apple, Add milk) or 'Done' to checkout.")

done = False
while not done:
    line = input("> ").strip()

    for command in line.split(","):
        command = command.strip().lower()

        if command == "done":
            done = True
            break

        parts = command.split(maxsplit=1)
        if len(parts) != 2 or parts[0] not in ("add", "remove"):
            print(f"Invalid command: '{command}'. Use 'Add <item>', 'Remove <item>', or 'Done'.")
            continue

        action, item = parts[0], parts[1].strip("'\" ")

        if action == "add":
            if item in store:
                cart.append(item)
                print(f"Added {item}.")
            else:
                print(f"Error: '{item}' is not in the store.")
        else:
            if item in cart:
                cart.remove(item)
                print(f"Removed {item}.")
            else:
                print(f"Error: '{item}' is not in your cart.")

total = sum(store[item] for item in cart)
print(f"Total Bill: ${total}")
```

# Question 13

```python
numbers = list(map(int, input("Enter numbers separated by spaces: ").split()))

peaks = []
for i in range(1, len(numbers) - 1):
    if numbers[i] > numbers[i - 1] and numbers[i] > numbers[i + 1]:
        peaks.append(numbers[i])

print("Peaks:", peaks)
```

# Question 14

```python
original = {'A': 1, 'B': 2, 'C': 1}

flipped = {}
for key, value in original.items():
    if value not in flipped:
        flipped[value] = []
    flipped[value].append(key)

print(flipped)
```

# Question 15

```python
text = input("Enter a string: ")

vowels = "aeiou"
result = {}

for index, ch in enumerate(text):
    if ch.lower() in vowels:
        result.setdefault(ch.lower(), []).append(index)

print(result)
```
